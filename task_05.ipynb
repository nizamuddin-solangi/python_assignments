```python
def calculator(expression):

    numbers = []
    operators = []

    def calculate():
        b = numbers.pop()
        a = numbers.pop()
        op = operators.pop()

        if op == "+":
            numbers.append(a + b)
        elif op == "-":
            numbers.append(a - b)
        elif op == "*":
            numbers.append(a * b)
        elif op == "/":
            if b == 0:
                raise ValueError("Division by zero!")
            numbers.append(a / b)

    def priority(op):
        if op == "+" or op == "-":
            return 1
        if op == "*" or op == "/":
            return 2
        return 0

    i = 0

    while i < len(expression):

        char = expression[i]

        if char.isspace():
            i += 1
            continue

        # Read numbers
        if char.isdigit():
            number = ""

            while i < len(expression) and expression[i].isdigit():
                number += expression[i]
                i += 1

            numbers.append(int(number))
            continue

        # Handle unary minus
        if char == "-" and (
            i == 0 or expression[i - 1] in "+-*/("
        ):
            numbers.append(0)

        # Opening bracket
        if char == "(":
            operators.append(char)

        # Closing bracket
        elif char == ")":
            while operators and operators[-1] != "(":
                calculate()

            if not operators:
                raise ValueError("Invalid parentheses!")

            operators.pop()

        # Operators
        elif char in "+-*/":

            while (
                operators
                and operators[-1] != "("
                and priority(operators[-1]) >= priority(char)
            ):
                calculate()

            operators.append(char)

        else:
            raise ValueError("Invalid character!")

        i += 1

    # Finish remaining calculations
    while operators:
        if operators[-1] == "(":
            raise ValueError("Invalid parentheses!")

        calculate()

    if len(numbers) != 1:
        raise ValueError("Invalid expression!")

    return numbers[0]


# Test
print(calculator("-(3 + 2) * 4 + 18 / 3"))
```
