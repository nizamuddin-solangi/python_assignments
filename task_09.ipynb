```python
def read_errors(path):
    read_errors.bad_lines = 0

    with open(path, "r") as file:
        for number, line in enumerate(file, start=1):

            try:
                parts = line.strip().split("|")

                if len(parts) != 3:
                    raise ValueError()

                timestamp = parts[0].strip()
                level = parts[1].strip()
                message = parts[2].strip()

                from datetime import datetime
                datetime.strptime(timestamp, "%Y-%m-%d %H:%M:%S")

                if not level or not message:
                    raise ValueError()

                if level == "ERROR":
                    yield (number, timestamp, message)

            except ValueError:
                read_errors.bad_lines += 1


def summarize_errors(path):
    summary = {}

    for number, timestamp, message in read_errors(path):
        date = timestamp[:10]

        if date not in summary:
            summary[date] = 0

        summary[date] += 1

    print("Errors:", summary)
    print("Malformed lines:", read_errors.bad_lines)


# Test
summarize_errors("app.log")
```
