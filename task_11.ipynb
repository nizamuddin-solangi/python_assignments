```python
import csv

students = {}

# Read students
with open("students.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        students[row["student_id"]] = {
            "name": row["name"],
            "marks": []
        }

# Read scores
seen = set()

with open("scores.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        sid = row["student_id"]
        subject = row["subject"]
        marks = float(row["marks"])

        if sid not in students:
            print("Unknown student:", sid)
            continue

        if not 0 <= marks <= 100:
            print("Invalid marks:", sid)
            continue

        if (sid, subject) in seen:
            print("Duplicate subject:", sid, subject)
            continue

        seen.add((sid, subject))
        students[sid]["marks"].append(marks)

report = []

for sid, data in students.items():

    if data["marks"]:
        average = sum(data["marks"]) / len(data["marks"])
    else:
        average = 0

    if average >= 85:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 55:
        grade = "C"
    elif average >= 40:
        grade = "D"
    else:
        grade = "F"

    report.append({
        "student_id": sid,
        "name": data["name"],
        "average": average,
        "grade": grade
    })

report.sort(key=lambda x: (-x["average"], x["student_id"]))

for i, student in enumerate(report):
    student["rank"] = i + 1

with open("report.csv", "w", newline="") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["student_id", "name", "average", "grade", "rank"]
    )

    writer.writeheader()
    writer.writerows(report)

print("Gradebook created!")
```
