```python
from abc import ABC, abstractmethod


# Parent abstract class
class Employee(ABC):

    def __init__(self, name):
        self.name = name

    @abstractmethod
    def calculate_pay(self):
        pass


# Salaried employee
class SalariedEmployee(Employee):

    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

    def calculate_pay(self):
        return self.salary


# Hourly employee
class HourlyEmployee(Employee):

    def __init__(self, name, rate, hours):
        super().__init__(name)
        self.rate = rate
        self.hours = hours

    def calculate_pay(self):
        if self.hours > 40:
            return (40 * self.rate) + ((self.hours - 40) * self.rate * 1.5)

        return self.hours * self.rate


# Commission employee
class CommissionEmployee(Employee):

    def __init__(self, name, base, sales, rate):
        super().__init__(name)
        self.base = base
        self.sales = sales
        self.rate = rate

    def calculate_pay(self):
        return self.base + (self.sales * self.rate)


# Create employees
employees = [
    SalariedEmployee("Ali", 3000),
    HourlyEmployee("Sara", 20, 45),
    CommissionEmployee("Ahmed", 500, 4000, 0.10)
]


# Payroll summary
print("EMPLOYEE PAYROLL SUMMARY")

for employee in employees:
    print(employee.name, "=", employee.calculate_pay())
```
