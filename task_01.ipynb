```python
transactions = [
    {"id": 1, "customer": "Ali", "category": "Food", "amount": 500, "status": "success"},
    {"id": 2, "customer": "Ali", "category": "Travel", "amount": 200, "status": "failed"},
    {"id": 3, "customer": "Sara", "category": "Food", "amount": 700, "status": "success"}
]

customers = ["Ali", "Sara", "Ahmed"]

report = {}

# Calculate spending
for name in customers:
    report[name] = {"total": 0, "category": {}}

for t in transactions:
    if t["status"] == "success":
        name = t["customer"]

        if t["id"] not in [x["id"] for x in transactions[:transactions.index(t)]]:
            report[name]["total"] += t["amount"]

            cat = t["category"]
            if cat not in report[name]["category"]:
                report[name]["category"][cat] = 0

            report[name]["category"][cat] += t["amount"]

# Create final report
result = []

for name in customers:
    data = report[name]

    if data["category"]:
        top = max(data["category"], key=data["category"].get)
    else:
        top = None

    result.append({
        "customer": name,
        "total": data["total"],
        "top_category": top
    })

# Sort spending from highest to lowest
result.sort(key=lambda x: (-x["total"], x["customer"]))

# Add ranks
for i in range(len(result)):
    result[i]["rank"] = i + 1

print(result)
```
