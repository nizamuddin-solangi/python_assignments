```python
def longest_unique_window(text):

    start = 0
    longest = ""
    best_start = 0
    seen = {}

    for end in range(len(text)):

        char = text[end]

        if char in seen and seen[char] >= start:
            start = seen[char] + 1

        seen[char] = end

        if end - start + 1 > len(longest):
            longest = text[start:end + 1]
            best_start = start

    return (longest, best_start, len(longest))


# Test
print(longest_unique_window("abcabcbb"))
print(longest_unique_window(""))
print(longest_unique_window("pwwkew"))
print(longest_unique_window("a b!a"))
```
