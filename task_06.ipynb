```python
from dataclasses import dataclass


# Immutable transaction record
@dataclass(frozen=True)
class Transaction:
    transaction_id: int
    action: str
    amount: float
    balance: float


# Account class
class Account:

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance
        self.transactions = []

    @property
    def balance(self):
        return self.__balance

    def deposit(self, amount, transaction_id):
        if amount <= 0:
            raise ValueError("Amount must be positive!")

        self.__balance += amount

        self.transactions.append(
            Transaction(transaction_id, "Deposit", amount, self.__balance)
        )

    def withdraw(self, amount, transaction_id):
        if amount <= 0:
            raise ValueError("Amount must be positive!")

        if amount > self.__balance:
            raise ValueError("Insufficient funds!")

        self.__balance -= amount

        self.transactions.append(
            Transaction(transaction_id, "Withdrawal", amount, self.__balance)
        )

    def statement(self):
        for transaction in self.transactions:
            print(transaction)


# Bank class
class Bank:

    def __init__(self):
        self.accounts = {}
        self.transaction_id = 1

    def create_account(self, name, balance):
        if balance < 0:
            raise ValueError("Invalid balance!")

        self.accounts[name] = Account(name, balance)

    def get_account(self, name):
        if name not in self.accounts:
            raise ValueError("Account does not exist!")

        return self.accounts[name]

    def deposit(self, name, amount):
        account = self.get_account(name)
        account.deposit(amount, self.transaction_id)
        self.transaction_id += 1

    def withdraw(self, name, amount):
        account = self.get_account(name)
        account.withdraw(amount, self.transaction_id)
        self.transaction_id += 1

    def transfer(self, sender, receiver, amount):

        if amount <= 0:
            raise ValueError("Amount must be positive!")

        account1 = self.get_account(sender)
        account2 = self.get_account(receiver)

        if account1.balance < amount:
            raise ValueError("Insufficient funds!")

        # Transfer money
        account1.withdraw(amount, self.transaction_id)
        self.transaction_id += 1

        account2.deposit(amount, self.transaction_id)
        self.transaction_id += 1


# Create bank
bank = Bank()

# Create accounts
bank.create_account("A", 1000)
bank.create_account("B", 200)

# Transfer 350
bank.transfer("A", "B", 350)

# Display balances
print("Account A:", bank.get_account("A").balance)
print("Account B:", bank.get_account("B").balance)

# Display statements
print("\nAccount A Statement:")
bank.get_account("A").statement()

print("\nAccount B Statement:")
bank.get_account("B").statement()
```
