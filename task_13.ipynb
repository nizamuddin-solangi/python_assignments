```python
from datetime import datetime

class BookingScheduler:

    def __init__(self):
        self.bookings = []

    def add_booking(self, room, start, end, booking_id):

        if end <= start:
            return "Invalid booking time"

        for booking in self.bookings:

            if booking["id"] == booking_id:
                return "Duplicate booking ID"

            if booking["room"] == room:

                if start < booking["end"] and end > booking["start"]:
                    return "Booking conflict"

        self.bookings.append({
            "room": room,
            "start": start,
            "end": end,
            "id": booking_id
        })

        self.bookings.sort(key=lambda x: (x["room"], x["start"]))

        return "Booking accepted"


# Test
scheduler = BookingScheduler()

time1 = datetime(2026, 11, 7, 9, 0)
time2 = datetime(2026, 11, 7, 10, 0)
time3 = datetime(2026, 11, 7, 11, 0)

print(scheduler.add_booking("A", time1, time2, 1))
print(scheduler.add_booking("A", time2, time3, 2))

print(scheduler.add_booking(
    "A",
    datetime(2026, 11, 7, 9, 30),
    datetime(2026, 11, 7, 10, 30),
    3
))
```
