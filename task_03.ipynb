```python
def find_paths(data, target):
    paths = []

    def search(item, path):

        if type(item) == dict:
            for key, value in item.items():
                if path == "":
                    new_path = key
                else:
                    new_path = path + "." + key

                search(value, new_path)

        elif type(item) == list:
            for i in range(len(item)):
                new_path = path + "[" + str(i) + "]"
                search(item[i], new_path)

        else:
            if type(item) == type(target) and item == target:
                paths.append(path)

    search(data, "")

    return paths


# Test
data = {
    "a": [
        {"x": 5},
        5
    ],
    "b": {
        "y": 5
    }
}

print(find_paths(data, 5))
```
