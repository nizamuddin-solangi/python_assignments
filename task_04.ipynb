```python
def execution_planner(tasks, dependencies):

    graph = {}
    count = {}

    for task in tasks:
        graph[task] = []   
        count[task] = 0   

    for a, b in dependencies:
        if a not in graph or b not in graph:
            raise ValueError("Unknown task!")

        graph[a].append(b)
        count[b] += 1

    result = []
    available = []

    for task in tasks:
        if count[task] == 0:
            available.append(task)

    while available:
        available.sort()
        task = available.pop(0)

        result.append(task)

        for next_task in graph[task]:
            count[next_task] -= 1

            if count[next_task] == 0:
                available.append(next_task)

    if len(result) != len(tasks):
        raise ValueError("Cycle detected!")

    return result


# Example
tasks = ["build", "test", "deploy"]

dependencies = [
    ("build", "test"),
    ("test", "deploy")
]

print(execution_planner(tasks, dependencies))
```
